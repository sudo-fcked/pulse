```python
# Initial exploration
import pandas as pd
import numpy as np

# Load April 2026 sample
csv_path = "../data/raw/MBTA_Bus_Arrival_Departure_Times_2026/MBTA-Bus-Arrival-Departure-Times_2026-04.csv"
df = pd.read_csv(csv_path)

print(f"Total rows: {len(df):,}")
print("\nColumns and missing counts:")
print(df.isnull().sum())

# Examine timestamps
print("\nSample scheduled values:", df['scheduled'].dropna().head(3).tolist())
print("Sample actual values:   ", df['actual'].dropna().head(3).tolist())

# Convert and calculate delay
sched = pd.to_datetime(df['scheduled'])
act = pd.to_datetime(df['actual'])
delays = (act - sched).dt.total_seconds() / 60.0

print("\nDelay summary stats (unfiltered minutes):")
print(delays.describe())

# Filter -30 to +120 minutes and compute late rate (> 3 minutes)
valid_mask = (delays >= -30) & (delays <= 120)
valid_delays = delays[valid_mask]
late_count = (valid_delays > 3.0).sum()

print(f"\nValid rows after cleaning: {len(valid_delays):,} (dropped {len(df) - len(valid_delays):,})")
print(f"Late rate (>3 min delay): {late_count / len(valid_delays) * 100:.2f}%")
```
